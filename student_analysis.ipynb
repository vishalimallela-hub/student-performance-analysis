```python
import matplotlib.pyplot as plt 
```

```python
failure_avg = df.groupby("failures")["G3"].mean()
```

```python
failure_avg.plot(kind="bar")
plt.xlabel("Number of past failures")
plt.ylabel("Average G3")
plt.title("Past Failures vs Average Final Grade")
plt.show()
```

```python
failure_avg.plot(kind="bar", figsize=(7,5))
plt.xlabel("Number of past failures")
plt.ylabel("Average Final Grade (G3)")
plt.title("Past Failures vs Average Final Grade")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../visualizations/failures_vs_g3.png", dpi=300)
plt.show()
```

```python
studytime_avg = df.groupby("studytime")["G3"].mean()
```

```python
studytime_avg.plot(kind="bar", figsize=(7,5))
plt.xlabel("Studytime Category")
plt.ylabel("Average Final Grade (G3)")
plt.title("Studytime vs Average Final Grade")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../visualizations/studytime_vs_g3.png", dpi=300)
plt.show()
```

```python
absence_avg = df.groupby("absence_group")["G3"].mean()
```

```python
absence_avg.plot(kind="bar", figsize=(7,5))
plt.xlabel("Absence Group")
plt.ylabel("Average Final Grade (G3)")
plt.title("Absence Group vs Average Final Grade")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../visualizations/absence_group_vs_g3.png", dpi=300)
plt.show()
```

```python
school_avg = df.groupby("school")["G3"].mean()
```

```python
school_avg.plot(kind="bar", figsize=(7,5))
plt.xlabel("School")
plt.ylabel("Average Final Grade (G3)")
plt.title("School vs Average Final Grade")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../visualizations/school_vs_g3.png", dpi=300)
plt.show()
```
